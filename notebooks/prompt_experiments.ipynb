## Tableau comparatif réel

Après trois exécutions par variante, compléter le tableau avec moyenne et étendue. Ne pas mélanger le score de fixture avec ces mesures.

```python
variants = ['zero_shot', 'few_shot', 'deliberative']
rows = []
for variant in variants:
    for path in sorted((ROOT / 'outputs' / 'runs').glob(f'{variant}_scores_*.json')):
        report = json.loads(path.read_text(encoding='utf-8'))
        rows.append({
            'variant': variant,
            'run': path.stem,
            **report['metrics'],
        })
rows
```
